```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("muthuj7/weather-dataset")

# Set the path to the file you'd like to load
file_path = f"{path}/weatherHistory.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print(f"Limpiar datos nulos o erróneos. Si los hay y extremos mayor de 95% distribución y menor 5% distribución.\n")
df = df[(df['Temperature (C)'] <= df['Temperature (C)'].quantile(0.95)) & (df['Temperature (C)'] >= df['Temperature (C)'].quantile(0.05))]
df = df[(df['Humidity'] <= df['Humidity'].quantile(0.95)) & (df['Humidity'] >= df['Humidity'].quantile(0.05))]

print("Crear columnas para temperaturas máximas, mínimas y promedios diarios. \n")
daily_temps = df.groupby('Formatted Date')['Temperature (C)'].agg(['max', 'min', 'mean']).reset_index()
daily_temps.columns = ['Formatted Date', 'Max Temperature (C)', 'Min Temperature (C)', 'Mean Temperature (C)']
print(daily_temps.head())
```
