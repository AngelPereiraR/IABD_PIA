```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("sidtwr/videogames-sales-dataset")

# Set the path to the file you'd like to load
file_path = f"{path}/Video_Games_Sales_as_at_22_Dec_2016.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("\nMedia del precio de la vivienda por ubicación:", f"{df.groupby('Platform')['Global_Sales'].mean()}")
```
