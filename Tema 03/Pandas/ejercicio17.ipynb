```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("fedesoriano/electric-power-consumption")

# Set the path to the file you'd like to load
file_path = f"{path}/powerconsumption.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("Crear una columna para consumo promedio diario y mensual.\n")
df['Datetime'] = pd.to_datetime(df['Datetime'])
df['Date'] = df['Datetime'].dt.date
df['Month'] = df['Datetime'].dt.to_period('M')
df['Consumption'] = pd.merge(df['PowerConsumption_Zone1'], df['PowerConsumption_Zone2'], left_index=True, right_index=True).mean(axis=1)
df['Consumption'] = pd.merge(df['Consumption'], df['PowerConsumption_Zone3'], left_index=True, right_index=True).mean(axis=1)
daily_avg = df.groupby('Date')['Consumption'].mean().reset_index()
monthly_avg = df.groupby('Month')['Consumption'].mean().reset_index()
print("Daily Average Consumption:\n", daily_avg.head())
print("Monthly Average Consumption:\n", monthly_avg.head())
```
