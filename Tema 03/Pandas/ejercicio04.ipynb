# Calcular la media de las temperaturas por ciudad
## Descripción:
Cargar un dataset de temperaturas históricas y calcular la media de la temperatura por ciudad.
## Fuente de datos:
Global Land Temperatures Datase

```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("vijayvvenkitesh/global-land-temperatures-by-country")

# Set the path to the file you'd like to load
file_path = f"{path}/GlobalLandTemperaturesByCountry.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("\nMean temperature by country:\n", df.groupby('Country')['AverageTemperature'].mean())
```
