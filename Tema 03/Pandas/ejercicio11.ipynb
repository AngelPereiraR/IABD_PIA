```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("harshitshankhdhar/imdb-dataset-of-top-1000-movies-and-tv-shows")

# Set the path to the file you'd like to load
file_path = f"{path}/imdb_top_1000.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("Películas de Drama:\n", df[df['Genre'].str.contains('Drama', na=False)])

print("Duración promedio de las películas por año de Drama:")
df['Runtime_Minutes'] = df['Runtime'].str.replace(' min', '').astype(int)
print(df[df['Genre'].str.contains('Drama', na=False)].groupby('Released_Year')['Runtime_Minutes'].mean())
```
