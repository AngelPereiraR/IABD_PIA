```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("brendan45774/test-file")

# Set the path to the file you'd like to load
file_path = f"{path}/tested.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("Rellenar los valores nulos en la columna de edad con la mediana por clase.")
median_ages = df.groupby('Pclass')['Age'].median()
for pclass, median_age in median_ages.items():
    df.loc[(df['Age'].isnull()) & (df['Pclass'] == pclass), 'Age'] = median_age

print("Agrupar por clase y calcular el porcentaje de supervivencia.\n", df.groupby('Pclass')['Survived'].mean())
```
