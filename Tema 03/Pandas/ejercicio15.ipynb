```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("yasserh/housing-prices-dataset")

# Set the path to the file you'd like to load
file_path = f"{path}/Housing.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("\nEliminar outliers en precios de viviendas. (más del 90% y menos del 10% por ejemplo):", f"{df.shape[0]} registros antes de limpiar.")
df = df[(df['price'] <= df['price'].quantile(0.90)) & (df['price'] >= df['price'].quantile(0.10))]
print(f"{df.shape[0]} registros después de limpiar.")
```
