```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("mathurtanvi/video-game-sales-dataset")

# Set the path to the file you'd like to load
file_path = f"{path}/Video game sales.xlsx"

# Load the Excel with the correct encoding
df = pd.read_excel(file_path)

print("First 5 records:\n", df.head())

print("Agrupar ventas por región y plataforma:\n", df.groupby(['platform'])[['north_american_sales', 'european_sales', 'japan_sales', 'aisan_sales']].sum())

print("Tabla pivote para ver las ventas de cada plataforma en diferentes años:\n", pd.pivot_table(df, values=['global_sales'], index=['platform'], columns=['release_year'], aggfunc='sum', fill_value=0))
```
