```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("zafarali27/car-price-prediction")

# Set the path to the file you'd like to load
file_path = f"{path}/car_price_prediction_.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("\nMedia del precio del coche por tipo de combustible:", f"{df.groupby('Fuel Type')['Price'].mean()}")
```
