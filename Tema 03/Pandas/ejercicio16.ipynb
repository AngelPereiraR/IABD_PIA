```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("khushikyad001/air-quality-data")

# Set the path to the file you'd like to load
file_path = f"{path}/AirQualityData.csv"

# Load the CSV with the correct encoding
df = pd.read_csv(file_path, encoding='latin-1')

print("First 5 records:\n", df.head())

print("\nManejar valores nulos y rellenar datos faltantes:", f"{df.shape[0]} registros antes de limpiar.")
df = df.dropna(subset=['PM2.5'])
print(f"{df.shape[0]} registros después de limpiar.")
df['PM2.5'].fillna(df['PM2.5'].median(), inplace=True)
print("\nValores nulos en PM2.5 después de rellenar:", f"{df['PM2.5'].isnull().sum()}")

print("Agrupar por fecha y crear una tabla pivote con los niveles de contaminación por fecha.\n")
pivot_table = df.pivot_table(values='PM2.5', index='Date', columns='Hour', aggfunc='mean')
print(pivot_table.head())
```
