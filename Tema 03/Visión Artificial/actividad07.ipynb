# Actividad 7 – Marcado manual de objeto

**Objetivo**: dibujar sobre imágenes.

En esta actividad vamos a:
1. Cargar una imagen
2. Dibujar una línea, un rectángulo y un círculo
3. Escribir un texto descriptivo
4. Guardar el resultado

## 1. Importar librerías necesarias

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Configurar matplotlib para mostrar imágenes en el notebook
%matplotlib inline
```

## 2. Cargar una imagen

Cambia la ruta por la ubicación de tu imagen.

```python
# Cargar la imagen
# Cambia 'imagen.jpg' por la ruta de tu imagen
imagen = cv2.imread('imagen.webp')

# Verificar que la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica la ruta del archivo.")
else:
    print("✓ Imagen cargada correctamente")
    
    # Crear una copia para dibujar (no modificar la original)
    imagen_dibujo = imagen.copy()
    
    # Mostrar la imagen original (convertir de BGR a RGB para matplotlib)
    imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(imagen_rgb)
    plt.axis('off')
    plt.title('Imagen original')
    plt.show()
```

## 3. Dibujar una línea

Usamos `cv2.line()` para dibujar una línea entre dos puntos.

```python
# Dibujar una línea
# Parámetros: imagen, punto_inicial, punto_final, color_BGR, grosor
punto_inicial = (50, 50)
punto_final = (300, 150)
color_linea = (0, 255, 0)  # Verde en BGR
grosor_linea = 3

cv2.line(imagen_dibujo, punto_inicial, punto_final, color_linea, grosor_linea)

print(f"✓ Línea dibujada desde {punto_inicial} hasta {punto_final}")
print(f"  Color (BGR): {color_linea}")
print(f"  Grosor: {grosor_linea} px")
```

## 4. Dibujar un rectángulo

Usamos `cv2.rectangle()` para dibujar un rectángulo.

```python
# Dibujar un rectángulo
# Parámetros: imagen, esquina_superior_izq, esquina_inferior_der, color_BGR, grosor (-1 para rellenar)
esquina_sup_izq = (400, 100)
esquina_inf_der = (600, 250)
color_rectangulo = (255, 0, 0)  # Azul en BGR
grosor_rectangulo = 2

cv2.rectangle(imagen_dibujo, esquina_sup_izq, esquina_inf_der, color_rectangulo, grosor_rectangulo)

print(f"✓ Rectángulo dibujado desde {esquina_sup_izq} hasta {esquina_inf_der}")
print(f"  Color (BGR): {color_rectangulo}")
print(f"  Grosor: {grosor_rectangulo} px")
```

## 5. Dibujar un círculo

Usamos `cv2.circle()` para dibujar un círculo.

```python
# Dibujar un círculo
# Parámetros: imagen, centro, radio, color_BGR, grosor (-1 para rellenar)
centro_circulo = (150, 300)
radio = 80
color_circulo = (0, 0, 255)  # Rojo en BGR
grosor_circulo = 3

cv2.circle(imagen_dibujo, centro_circulo, radio, color_circulo, grosor_circulo)

print(f"✓ Círculo dibujado en {centro_circulo}")
print(f"  Radio: {radio} px")
print(f"  Color (BGR): {color_circulo}")
print(f"  Grosor: {grosor_circulo} px")
```

## 6. Escribir texto descriptivo

Usamos `cv2.putText()` para escribir texto sobre la imagen.

```python
# Escribir texto
# Parámetros: imagen, texto, posición, fuente, tamaño, color_BGR, grosor
texto = "Imagen Marcada"
posicion_texto = (50, 450)
fuente = cv2.FONT_HERSHEY_SIMPLEX
tamano_fuente = 1.5
color_texto = (255, 255, 255)  # Blanco en BGR
grosor_texto = 3

cv2.putText(imagen_dibujo, texto, posicion_texto, fuente, tamano_fuente, color_texto, grosor_texto)

print(f"✓ Texto '{texto}' escrito en {posicion_texto}")
print(f"  Fuente: HERSHEY_SIMPLEX")
print(f"  Tamaño: {tamano_fuente}")
print(f"  Color (BGR): {color_texto}")

# Mostrar imagen con todos los dibujos
imagen_dibujo_rgb = cv2.cvtColor(imagen_dibujo, cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

# Original
axes[0].imshow(imagen_rgb)
axes[0].set_title('Original', fontsize=14, fontweight='bold')
axes[0].axis('off')

# Con marcado
axes[1].imshow(imagen_dibujo_rgb)
axes[1].set_title('Con Marcado Manual', fontsize=14, fontweight='bold')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## 7. Guardar el resultado

```python
# Guardar la imagen con los dibujos
nombre_resultado = 'imagen_marcada.jpg'
resultado = cv2.imwrite(nombre_resultado, imagen_dibujo)

if resultado:
    print(f"✓ Imagen marcada guardada como: {nombre_resultado}")
else:
    print(f"✗ Error al guardar la imagen")
```
