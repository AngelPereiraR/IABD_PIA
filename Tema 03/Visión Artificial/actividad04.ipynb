# Actividad 4 – Redimensionar sin deformar

**Objetivo**: mantener proporción al redimensionar.

En esta actividad vamos a:
1. Cargar una imagen
2. Redimensionarla a ancho 300 píxeles manteniendo proporción
3. Mostrar original y redimensionada

## 1. Importar librerías necesarias

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Configurar matplotlib para mostrar imágenes en el notebook
%matplotlib inline
```

## 2. Cargar una imagen

Cambia la ruta por la ubicación de tu imagen. Puedes usar cualquier imagen (jpg, png, etc.)

```python
# Cargar la imagen
# Cambia 'imagen.jpg' por la ruta de tu imagen
imagen = cv2.imread('imagen.webp')

# Verificar que la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica la ruta del archivo.")
else:
    print("✓ Imagen cargada correctamente")
    
    # Mostrar la imagen (convertir de BGR a RGB para matplotlib)
    imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(imagen_rgb)
    plt.axis('off')
    plt.title('Imagen cargada')
    plt.show()
```

## 3. Redimensionar manteniendo proporción

Para redimensionar sin deformar, calculamos el alto proporcionalmente basándonos en el nuevo ancho.

```python
# Obtener dimensiones originales
alto_original, ancho_original = imagen.shape[:2]

# Definir nuevo ancho
nuevo_ancho = 300

# Calcular nuevo alto manteniendo la proporción
proporcion = nuevo_ancho / ancho_original
nuevo_alto = int(alto_original * proporcion)

print("=" * 50)
print("DIMENSIONES")
print("=" * 50)
print(f"Original:           {ancho_original} x {alto_original} píxeles")
print(f"Redimensionada:     {nuevo_ancho} x {nuevo_alto} píxeles")
print(f"Proporción:         {proporcion:.4f}")
print("=" * 50)

# Redimensionar la imagen
imagen_redimensionada = cv2.resize(imagen, (nuevo_ancho, nuevo_alto), interpolation=cv2.INTER_AREA)

print("\n✓ Imagen redimensionada correctamente")
```

## 4. Mostrar original y redimensionada

```python
# Convertir ambas imágenes de BGR a RGB para matplotlib
imagen_redimensionada_rgb = cv2.cvtColor(imagen_redimensionada, cv2.COLOR_BGR2RGB)

# Mostrar las dos versiones
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Imagen original
axes[0].imshow(imagen_rgb)
axes[0].set_title(f'Original\n{ancho_original} x {alto_original} px', fontsize=12, fontweight='bold')
axes[0].axis('off')

# Imagen redimensionada
axes[1].imshow(imagen_redimensionada_rgb)
axes[1].set_title(f'Redimensionada\n{nuevo_ancho} x {nuevo_alto} px', fontsize=12, fontweight='bold')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
