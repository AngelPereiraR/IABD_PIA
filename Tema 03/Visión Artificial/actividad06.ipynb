# Actividad 6 – Ajuste de brillo y contraste

**Objetivo**: usar convertScaleAbs.

En esta actividad vamos a:
1. Cargar una imagen
2. Crear tres versiones: más brillante, más contraste y oscurecida
3. Mostrar las cuatro imágenes juntas

## 1. Importar librerías necesarias

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Configurar matplotlib para mostrar imágenes en el notebook
%matplotlib inline
```

## 2. Cargar una imagen

Cambia la ruta por la ubicación de tu imagen.

```python
# Cargar la imagen
# Cambia 'imagen.jpg' por la ruta de tu imagen
imagen = cv2.imread('imagen.webp')

# Verificar que la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica la ruta del archivo.")
else:
    print("✓ Imagen cargada correctamente")
    
    # Mostrar la imagen (convertir de BGR a RGB para matplotlib)
    imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(imagen_rgb)
    plt.axis('off')
    plt.title('Imagen cargada')
    plt.show()
```

## 3. Crear versión más brillante

`cv2.convertScaleAbs()` permite ajustar el brillo (beta) y contraste (alpha) de una imagen.
Fórmula: `nueva_imagen = alpha * imagen + beta`

```python
# Crear versión más brillante
# alpha = 1.0 (sin cambio de contraste), beta = 50 (aumentar brillo)
alpha_brillo = 1.0
beta_brillo = 50

imagen_brillante = cv2.convertScaleAbs(imagen, alpha=alpha_brillo, beta=beta_brillo)

print(f"✓ Versión más brillante creada")
print(f"  Alpha (contraste): {alpha_brillo}")
print(f"  Beta (brillo):     +{beta_brillo}")
```

## 4. Crear versión con más contraste

Aumentar alpha incrementa el contraste de la imagen.

```python
# Crear versión con más contraste
# alpha = 1.5 (aumentar contraste), beta = 0 (sin cambio de brillo)
alpha_contraste = 1.5
beta_contraste = 0

imagen_contraste = cv2.convertScaleAbs(imagen, alpha=alpha_contraste, beta=beta_contraste)

print(f"✓ Versión con más contraste creada")
print(f"  Alpha (contraste): {alpha_contraste}")
print(f"  Beta (brillo):     {beta_contraste}")
```

## 5. Crear versión oscurecida

Reducir beta oscurece la imagen (valores negativos reducen el brillo).

```python
# Crear versión oscurecida
# alpha = 1.0 (sin cambio de contraste), beta = -50 (reducir brillo)
alpha_oscura = 1.0
beta_oscura = -50

imagen_oscura = cv2.convertScaleAbs(imagen, alpha=alpha_oscura, beta=beta_oscura)

print(f"✓ Versión oscurecida creada")
print(f"  Alpha (contraste): {alpha_oscura}")
print(f"  Beta (brillo):     {beta_oscura}")
```

## 6. Mostrar las cuatro imágenes juntas

```python
# Convertir todas las imágenes de BGR a RGB para matplotlib
imagen_brillante_rgb = cv2.cvtColor(imagen_brillante, cv2.COLOR_BGR2RGB)
imagen_contraste_rgb = cv2.cvtColor(imagen_contraste, cv2.COLOR_BGR2RGB)
imagen_oscura_rgb = cv2.cvtColor(imagen_oscura, cv2.COLOR_BGR2RGB)

# Mostrar las cuatro versiones en una cuadrícula 2x2
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Imagen original
axes[0, 0].imshow(imagen_rgb)
axes[0, 0].set_title('Original', fontsize=14, fontweight='bold')
axes[0, 0].axis('off')

# Imagen más brillante
axes[0, 1].imshow(imagen_brillante_rgb)
axes[0, 1].set_title(f'Más Brillante (β=+{beta_brillo})', fontsize=14, fontweight='bold')
axes[0, 1].axis('off')

# Imagen con más contraste
axes[1, 0].imshow(imagen_contraste_rgb)
axes[1, 0].set_title(f'Más Contraste (α={alpha_contraste})', fontsize=14, fontweight='bold')
axes[1, 0].axis('off')

# Imagen oscurecida
axes[1, 1].imshow(imagen_oscura_rgb)
axes[1, 1].set_title(f'Oscurecida (β={beta_oscura})', fontsize=14, fontweight='bold')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```
