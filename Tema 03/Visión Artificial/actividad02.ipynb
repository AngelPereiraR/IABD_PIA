# Actividad 2 – Explorando una imagen

**Objetivo**: entender propiedades básicas de una imagen.

En esta actividad vamos a:
1. Cargar una imagen
2. Mostrar sus propiedades (alto, ancho, número de canales, tipo de dato)
3. Imprimir el valor de un píxel concreto
4. Guardar una copia con otro nombre

## 1. Importar librerías necesarias

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Configurar matplotlib para mostrar imágenes en el notebook
%matplotlib inline
```

## 2. Cargar una imagen

Cambia la ruta por la ubicación de tu imagen. Puedes usar cualquier imagen (jpg, png, etc.)

```python
# Cargar la imagen
# Cambia 'imagen.jpg' por la ruta de tu imagen
imagen = cv2.imread('imagen.webp')

# Verificar que la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica la ruta del archivo.")
else:
    print("✓ Imagen cargada correctamente")
    
    # Mostrar la imagen (convertir de BGR a RGB para matplotlib)
    imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(imagen_rgb)
    plt.axis('off')
    plt.title('Imagen cargada')
    plt.show()
```

## 3. Mostrar propiedades de la imagen

Vamos a explorar las propiedades básicas: alto, ancho, número de canales y tipo de dato.

```python
# Obtener dimensiones de la imagen
alto, ancho, canales = imagen.shape

print("=" * 50)
print("PROPIEDADES DE LA IMAGEN")
print("=" * 50)
print(f"Alto (height):        {alto} píxeles")
print(f"Ancho (width):        {ancho} píxeles")
print(f"Número de canales:    {canales}")
print(f"Tipo de dato:         {imagen.dtype}")
print(f"Tamaño total:         {imagen.size} elementos")
print(f"Forma (shape):        {imagen.shape}")
print("=" * 50)
```

## 4. Imprimir el valor de un píxel concreto

Vamos a acceder al valor de un píxel específico. Recuerda que en OpenCV, los colores están en formato BGR (Blue, Green, Red).

```python
# Coordenadas del píxel (fila, columna)
# Cambia estos valores para explorar diferentes píxeles
fila = 100
columna = 150

# Obtener el valor del píxel
pixel = imagen[fila, columna]

print("=" * 50)
print(f"VALOR DEL PÍXEL EN LA POSICIÓN ({fila}, {columna})")
print("=" * 50)
print(f"Valor completo:   {pixel}")
print(f"Canal Azul (B):   {pixel[0]}")
print(f"Canal Verde (G):  {pixel[1]}")
print(f"Canal Rojo (R):   {pixel[2]}")
print("=" * 50)

# Visualizar la ubicación del píxel en la imagen
imagen_con_punto = imagen_rgb.copy()
cv2.circle(imagen_con_punto, (columna, fila), 5, (255, 0, 0), -1)
plt.figure(figsize=(10, 6))
plt.imshow(imagen_con_punto)
plt.plot(columna, fila, 'r*', markersize=15)
plt.title(f'Ubicación del píxel ({fila}, {columna})')
plt.axis('off')
plt.show()
```

## 5. Guardar una copia con otro nombre

```python
# Guardar una copia de la imagen con otro nombre
nombre_copia = 'imagen_copia.jpg'

# Guardar la imagen
resultado = cv2.imwrite(nombre_copia, imagen)

if resultado:
    print(f"✓ Imagen guardada correctamente como: {nombre_copia}")
else:
    print(f"✗ Error al guardar la imagen")
```
