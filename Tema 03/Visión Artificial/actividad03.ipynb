# Actividad 3 – Conversión de color

**Objetivo**: trabajar con espacios de color.

En esta actividad vamos a:
1. Cargar una imagen en color
2. Convertirla a escala de grises
3. Convertirla a HSV
4. Mostrar las tres versiones
5. Guardar la versión en gris

## 1. Importar librerías necesarias

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Configurar matplotlib para mostrar imágenes en el notebook
%matplotlib inline
```

## 2. Cargar una imagen

Cambia la ruta por la ubicación de tu imagen. Puedes usar cualquier imagen (jpg, png, etc.)

```python
# Cargar la imagen
# Cambia 'imagen.jpg' por la ruta de tu imagen
imagen = cv2.imread('imagen.webp')

# Verificar que la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica la ruta del archivo.")
else:
    print("✓ Imagen cargada correctamente")
    
    # Mostrar la imagen (convertir de BGR a RGB para matplotlib)
    imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(imagen_rgb)
    plt.axis('off')
    plt.title('Imagen cargada')
    plt.show()
```

## 3. Conversión a escala de grises

OpenCV proporciona la función `cv2.cvtColor()` para convertir entre diferentes espacios de color.

```python
# Convertir la imagen a escala de grises
imagen_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

print("✓ Imagen convertida a escala de grises")
print(f"Forma original:     {imagen.shape}")
print(f"Forma en grises:    {imagen_gris.shape}")
print(f"Tipo de dato:       {imagen_gris.dtype}")
```

## 4. Conversión a HSV

HSV (Hue, Saturation, Value) es un espacio de color útil para tareas de detección y segmentación de colores.

```python
# Convertir la imagen a HSV
imagen_hsv = cv2.cvtColor(imagen, cv2.COLOR_BGR2HSV)

print("✓ Imagen convertida a HSV")
print(f"Forma HSV:          {imagen_hsv.shape}")
print(f"Tipo de dato:       {imagen_hsv.dtype}")
print("\nCanales HSV:")
print("  - H (Hue/Matiz):        Canal 0")
print("  - S (Saturation):       Canal 1")
print("  - V (Value/Valor):      Canal 2")
```

## 5. Mostrar las tres versiones y guardar la versión en gris

```python
# Mostrar las tres versiones
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Imagen original en RGB
axes[0].imshow(imagen_rgb)
axes[0].set_title('Original (RGB)', fontsize=14, fontweight='bold')
axes[0].axis('off')

# Imagen en escala de grises
axes[1].imshow(imagen_gris, cmap='gray')
axes[1].set_title('Escala de grises', fontsize=14, fontweight='bold')
axes[1].axis('off')

# Imagen HSV (convertir a RGB para visualización correcta)
imagen_hsv_rgb = cv2.cvtColor(imagen_hsv, cv2.COLOR_HSV2RGB)
axes[2].imshow(imagen_hsv_rgb)
axes[2].set_title('HSV', fontsize=14, fontweight='bold')
axes[2].axis('off')

plt.tight_layout()
plt.show()

# Guardar la versión en escala de grises
nombre_gris = 'imagen_gris.jpg'
resultado = cv2.imwrite(nombre_gris, imagen_gris)

if resultado:
    print(f"\n✓ Imagen en escala de grises guardada como: {nombre_gris}")
else:
    print(f"\n✗ Error al guardar la imagen")
```
