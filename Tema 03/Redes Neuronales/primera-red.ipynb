```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf
import numpy as np
```

# Obtención de los datos

```python
print ("Leyendo CSV")
temperature_df = pd.read_csv("celsius_a_fahrenheit.csv")

# Visualización
sns.scatterplot(x = temperature_df['Celsius'], y = temperature_df['Fahrenheit'])
plt.show()
```

# Selección de las variables independiente y objetivo

```python
# Cargando Set de Datos
print ("Seleccionando las columnas")
X_train = temperature_df['Celsius']
y_train = temperature_df['Fahrenheit']
```

# Creación del modelo

```python
# Creando el Modelo
print ("Creando el modelo")
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=10, input_shape=[1], activation='relu'))
model.add(tf.keras.layers.Dense(units=10, activation='relu'))
model.add(tf.keras.layers.Dense(units=5, activation='relu'))
model.add(tf.keras.layers.Dense(units=1))
```

# Compilación del modelo

```python
# Compilado
print ("Compilando el modelo")
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.6), loss='mean_squared_error')
```

# Entrenamiento del modelo

```python
# Entrenando el modelo
print ("Entrenando el modelo")
epochs_hist = model.fit(X_train, y_train, epochs = 100)
```

# Evaluación del modelo

```python
# Evaluando modelo
print ("Evaluando el modelo entrenado")
print ("Keys:")
print(epochs_hist.history.keys())
	
# Grafico
plt.plot(epochs_hist.history['loss'])
plt.title('Progreso de Pérdida durante Entrenamiento del Modelo')
plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.legend('Training Loss')
plt.show()
```

# Realizar una predicción utilizando el modelo entrenado

```python
Temp_C = 0
# Pasamos los datos como un arreglo
Temp_F = model.predict(np.array([Temp_C]))
print("Temperatura de Prediccion: " + str(Temp_F))
```
